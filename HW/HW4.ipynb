```python
import numpy as np
from collections import Counter

X = np.array([4, 3])
train = {'A': ((4, 4), 'Low'), 'B': ((5, 3), 'High'), 'C': ((3, 3), 'Low'), 'D': ((9, 9), 'High')}
```

```python
# a) Distances to all four training points
dists = sorted([(k, np.linalg.norm(X - np.array(pt)), lbl) for k, (pt, lbl) in train.items()], key=lambda x: x[1])
print("a) Distances:")
for k, d, _ in dists:
    print(f"   d(X, {k}) = {d:.4f}")
```

```python
# b) Three nearest points and their labels
nearest = dists[:3]
print("\nb) Three nearest (K=3):", [(k, lbl) for k, _, lbl in nearest])
```

```python
# c) Majority vote & effect of D
pred = Counter(lbl for _, _, lbl in nearest).most_common(1)[0][0]
print(f"\nc) Predicted class: {pred} (2 Low vs 1 High)")
print("   Would D affect prediction? No—D is the 4th farthest point (d = 7.81), so it is excluded when K = 3.")
```

```python
P_S, P_F_given_S, P_F_given_notS = 0.30, 0.80, 0.20

# a) Calculate P(notS)
P_notS = 1 - P_S
print(f"a) P(notS) = 1 - {P_S:.2f} = {P_notS:.2f}")
```

```python
# b) Numerator, denominator, and P(S | F)
num = P_F_given_S * P_S
den = num + (P_F_given_notS * P_notS)
P_S_given_F = num / den
print(f"\nb) Numerator   = {num:.2f}\n"
      f"   Denominator = {den:.2f}\n"
      f"   P(S | F)    = {num:.2f} / {den:.2f} = {P_S_given_F:.4f} ({P_S_given_F:.2%})")
```

```python
# c) Predicted class (threshold = 0.50)
pred = "Spam" if P_S_given_F >= 0.50 else "Not Spam"
print(f"\nc) Predicted class: {pred} (since {P_S_given_F:.4f} >= 0.50)")
```

```python
gini = lambda counts: 1 - sum((c / sum(counts)) ** 2 for c in counts)

# a) 6 High and 4 Low cars
ga = gini([6, 4])
print(f"a) Gini = 1 - (0.6^2 + 0.4^2) = {ga:.2f}")
```

```python
# b) 10 High and 0 Low cars
gb = gini([10, 0])
print(f"b) Gini = 1 - (1.0^2 + 0.0^2) = {gb:.2f}")
```

```python
# c) Comparison & splitting goal
print("c) Node (b) is purer (Gini = 0.00 vs 0.48). When splitting, a decision tree aims to minimize weighted Gini impurity (maximize purity) so child nodes contain more uniform class labels.")
```

```python
import pandas as pd

TP, FN, FP, TN = 18, 2, 7, 23
N = TP + FN + FP + TN

# a) Confusion matrix & total observations
cm = pd.DataFrame([[TP, FN], [FP, TN]], index=['Actual 1', 'Actual 0'], columns=['Predicted 1', 'Predicted 0'])
print(f"a) Confusion Matrix:\n{cm}\nTotal observations (N): {N}")
```

```python
# b) Metrics
acc = (TP + TN) / N
prec = TP / (TP + FP)
sens = TP / (TP + FN)
spec = TN / (TN + FP)
f1 = (2 * TP) / (2 * TP + FP + FN)

print(f"\nb) Accuracy    = {acc:.2%}\n"
      f"   Precision   = {prec:.2%}\n"
      f"   Sensitivity = {sens:.2%}\n"
      f"   Specificity = {spec:.2%}\n"
      f"   F1-score    = {f1:.2%}")
```

```python
# c) Metric directly affected by FN
print("\nc) Sensitivity (Recall) = TP / (TP + FN) is directly affected by false negatives (as well as Accuracy and F1-score).")
```

```python
# d) Comparison
print(f"\nd) Sensitivity ({sens:.2%}) is higher than Precision ({prec:.2%}). "
      "Sensitivity shows the model catches 90% of actual high-price cars, while Precision shows that only 72% of its positive predictions are actually high-price (due to 7 false alarms)")
```
