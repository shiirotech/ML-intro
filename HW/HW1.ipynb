```python
# y_hat = 92000 − 3500x1 − 2000x2 + 4000x3
# a) x1 = 4, x2 = 7, x3 = 0
# b) y_hat = 92000 - 3500 * 4 - 2000 * 7 + 4000 * 0 = 92000 - 14000 - 14000 + 0 = 64000
# c) Interpret: each extra year reduces predicted price by 3500 PLN, each extra 10000 km reduce predicted
#               price by 2000 PLN. Service history adds 4000 PLN relative to the same car without it.
# d) With full service history:
#    y_hat = 92000 - 3500 * 4 - 2000 * 7 + 4000 * 1 = 92000 - 14000 - 14000 + 4000 = 68000 (exactly 4000 PLN more)
```

```python
import numpy as np
import pandas as pd

cars = ['A', 'B', 'C', 'D']
actual = np.array([45, 60, 75, 90], dtype=int)
predicted = np.array([50, 58, 80, 85], dtype=int)

df = pd.DataFrame({
    'Car': cars,
    'Actual y': actual,
    'Predicted y_hat': predicted
})
```

```python
df
```

```python
df['Residual e = y - y_hat'] = df['Actual y'] - df['Predicted y_hat']
df['Prediction Status'] = np.where(
    df['Residual e = y - y_hat'] > 0, 'Underestimated',
    np.where(df['Residual e = y - y_hat'] < 0, 'Overestimated', 'Exact')
)

print(df[['Car', 'Actual y', 'Predicted y_hat', 'Residual e = y - y_hat', 'Prediction Status']])

underestimated = df.loc[df['Residual e = y - y_hat'] > 0, 'Car'].tolist()
print(f"\nUnderestimated cars: {', '.join(underestimated)}")
```

```python
df['Absolute Error |e|'] = np.abs(df['Residual e = y - y_hat'])
df['Squared Error e^2'] = df['Residual e = y - y_hat'] ** 2

print(df[['Car', 'Residual e = y - y_hat', 'Absolute Error |e|', 'Squared Error e^2']])
```

```python
n = len(df)
abs_errors = df['Absolute Error |e|']
sq_errors = df['Squared Error e^2']

sum_abs = abs_errors.sum()
mae = sum_abs / n

sum_sq = sq_errors.sum()
mse = sum_sq / n

rmse = np.sqrt(mse)

print("--- Step-by-Step Calculations ---")
print(f"1. MAE = ({' + '.join(abs_errors.astype(str))}) / {n} = {sum_abs} / {n} = {mae:.2f} thousand PLN")
print(f"2. MSE = ({' + '.join(sq_errors.astype(str))}) / {n} = {sum_sq} / {n} = {mse:.2f} (thousand PLN)^2")
print(f"3. RMSE = sqrt({mse:.2f}) = {rmse:.4f} thousand PLN (approx. {rmse * 1000:.2f} PLN)")
```

```python
max_abs_err = df['Absolute Error |e|'].max()
cars_with_max_err = df.loc[df['Absolute Error |e|'] == max_abs_err, 'Car'].tolist()

print(f"Maximum absolute prediction error: {max_abs_err} thousand PLN")
print(f"Cars tied for the largest absolute prediction error: {', '.join(cars_with_max_err)}")

# Explanation:
# Cars A, C and D all tie for the largest absolute prediction error of 5000 PLN
# (|45 - 50| = 5, |75 - 80| = 5, and |90 - 85| = 5), whereas Car B has an absolute error
# of only 2000 PLN (|60 - 58| = 2). Although Car D has a positive residual (+5)
# and Cars A and C have negative residuals (-5), absolute error measures magnitude without
# regard to direction, resulting in a three-way tie.
```
