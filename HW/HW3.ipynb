```python
import numpy as np

pts = {'A': (1, 2), 'B': (2, 2), 'C': (3, 3), 'D': (8, 8), 'E': (9, 8)}
mu1, mu2 = np.array([1, 2]), np.array([8, 8])

# a) Distances & b) Allocation
c1, c2 = [], []
print("a) Distances d = sqrt[(x_P - x_Q)^2 + (y_P - y_Q)^2]:")
for k, v in pts.items():
    p = np.array(v)
    d1, d2 = np.linalg.norm(p - mu1), np.linalg.norm(p - mu2)
    (c1 if d1 < d2 else c2).append(k)
    print(f"{k}{v}: d1 = {d1:.2f}, d2 = {d2:.2f}")

print(f"\nb) Clusters: C1 = {c1}, C2 = {c2}")
```

```python
# c) Updated centroids
mu1_new = np.mean([pts[k] for k in c1], axis=0)
mu2_new = np.mean([pts[k] for k in c2], axis=0)
print(f"\nc) Updated centroids:\n"
      f"mu1_new = ((1+2+3)/3, (2+2+3)/3) = ({mu1_new[0]:.2f}, {mu1_new[1]:.2f})\n"
      f"mu2_new = ((8+9)/2, (8+8)/2) = ({mu2_new[0]:.2f}, {mu2_new[1]:.2f})")
```

```python
# d) Reassign
c1_new = [k for k, v in pts.items() if np.linalg.norm(v - mu1_new) < np.linalg.norm(v - mu2_new)]
c2_new = [k for k in pts if k not in c1_new]
print(f"\nd) New clusters: C1 = {c1_new}, C2 = {c2_new}. Changed? {c1 != c1_new}")
```

```python
# Silhouette formula: s = (b - a) / max(a, b)
s = lambda a, b: (b - a) / max(a, b)
```

```python
# a) a = 3, b = 8
sa = s(3, 8)
print(f"a) s = (8 - 3) / max(3, 8) = 5 / 8 = {sa:.4f}")
print("   Sign & size: Positive and fairly close to 1, indicating the point is well-matched to its own cluster and clearly separated from the nearest neighboring cluster.")
```

```python
# b) a = 6, b = 4
sb = s(6, 4)
print(f"\nb) s = (4 - 6) / max(6, 4) = -2 / 6 = {sb:.4f}")
print("   Negative score: Suggests the point is closer on average to another cluster than its own (likely misclassified/assigned to the wrong cluster).")
```
