```python
# z = −1.20 + 1.00x1 − 0.60x2
# 1) x1 = 1, x2 = 0
#    z = -1.20 + 1.00 * 1 - 0.60 * 0 = -0.20
#    e^(-z) = e^(0.20) = 1.2214
#    p = 1 / (1 + 1.2214) = 0.4502 (45.02%)
#
# 2) Predicted class: 0 (Not high-price)
#    Why: Because p = 0.4502 < 0.50
#
# 3) x1 = 1, x2 = 1
#    z = -1.20 + 1.00(1) - 0.60(1) = -0.80
#    e^(-z) = e^(0.80) = 2.2255
#    p = 1 / (1 + 2.2255) = 0.3100 (31.00%)
#    Predicted class (threshold 0.50): 0 (Not high-price), since 0.3100 < 0.50
#
# 4) If we change threshold to be 0.4, car from example a) becomes high-priced, as 0.4502 > 0.4
```

```python
import pandas as pd

TP, FN, FP, TN = 12, 4, 6, 18

# a) Confusion Matrix
cm = pd.DataFrame([[TP, FN], [FP, TN]], index=['Actual 1', 'Actual 0'], columns=['Predicted 1', 'Predicted 0'])
print("a) Confusion Matrix:\n", cm)
```

```python
# b) Totals
N = TP + FN + FP + TN
actual_pos = TP + FN
print(f"\nb) Total cars (N): {N}, Actual positives (TP + FN): {actual_pos}")
```

```python
# c) Metrics
acc = (TP + TN) / N
prec = TP / (TP + FP)
rec = TP / (TP + FN)
spec = TN / (TN + FP)
f1 = (2 * TP) / (2 * TP + FP + FN)

print(f"\nc) Metrics:\n"
      f"Accuracy    = ({TP}+{TN})/{N} = {acc:.2%}\n"
      f"Precision   = {TP}/({TP}+{FP}) = {prec:.2%}\n"
      f"Sensitivity = {TP}/({TP}+{FN}) = {rec:.2%}\n"
      f"Specificity = {TN}/({TN}+{FP}) = {spec:.2%}\n"
      f"F1-score    = {2*TP}/({2*TP}+{FP}+{FN}) = {f1:.2%}\n")
```

```python
# d) Explanation
print("d) FN = 4 means 4 actually high-price cars were misclassified as not high-price. "
      "Accuracy alone hides whether errors come from missing high-price cars (FN) or false alarms (FP) "
      "and can be misleading when classes are imbalanced.")
```
